```python
import numpy as np
import sympy as sp

x, y = sp.symbols('x y')
sp.init_printing()
np.random.seed(42)
print('Setup complete. SymPy', sp.__version__, '| NumPy', np.__version__)
```

```python
def f(x):
    return x**2

h = 1e-6
slope_at_3 = (f(3 + h) - f(3)) / h
print('Numerical f\'(3):', round(slope_at_3, 4), ' (exact = 6)')
```

```python

expr = y**2+y*y
deriv = sp.diff(expr, y)
print('d/dy (y**2+y**y) =', deriv)
print('Symbolic f\'(3) =', deriv.subs(y, 3))
```

```python
def g(x):
    return x**3 + 2*x

# 1. Numerical derivative at x = 2 (use h = 1e-6)
h = 1e-6
slope_at_2 = (f(2 + h) - f(2)) / h
print('Numerical f\'(3):', round(slope_at_2, 4), ' (exact = 6)')

# 2. Symbolic derivative of x**3 + 2*x
expr = x**3 + 2*x
deriv = sp.diff(expr, x)
print('d/dx (x**3 + 2*x) =', deriv)

# 3. Evaluate the symbolic derivative at x = 2 and compare
print('Symbolic f\'(2) =', deriv.subs(x, 2))
```

```python
f2 = x**2 + 3*x*y + y**2


print('df/dx =', sp.diff(f2, x))
print('df/dy =', sp.diff(f2, y))
```

```python



grad = [sp.diff(f2, v) for v in (x, y)]
print('grad f =', grad)
grad_at = [g.subs({x: 1, y: 2}) for g in grad]
print('grad f at (1, 2) =', grad_at)
```

```python
h2 = x**2 * y + sp.sin(y)

# 1. dh/dx and dh/dy


print('dh/dx =', sp.diff(h2, x))
print('dh/dy =', sp.diff(h2, y))

# 2. Assemble the gradient list
grad = [sp.diff(h2, v) for v in (x, y)]
print('grad h =', grad)

# 3. Evaluate at (x=2, y=0)  -> hint: .subs({x: 2, y: 0})
grad_at = [g.subs({x: 2, y: 0}) for g in grad]
print('grad h at (2, 0) =', grad_at)
```

```python
by_hand = sp.cos(x**2) * 2*x
by_sympy = sp.diff(sp.sin(x**2), x)

print('By hand :', by_hand)
print('By SymPy:', by_sympy)
print('Match?  ', sp.simplify(by_hand - by_sympy) == 0)
```

```python
import numpy as np

Xb = np.random.randn(6, 4)
Yb = np.random.randn(6, 1)
Wa = np.random.randn(4, 8) * 0.1     # input -> hidden
Wb = np.random.randn(8, 1) * 0.1     # hidden -> output

# -----------------------------------------------------------
# 1. Forward pass
# -----------------------------------------------------------

z1 = Xb @ Wa
h = np.maximum(0, z1)
y_hat = h @ Wb

loss = np.mean((y_hat - Yb)**2)

print("Loss before:", loss)


dy = 2 * (y_hat - Yb) / Yb.shape[0]
dWb = h.T @ dy
dh = dy @ Wb.T
dz1 = dh * (z1 > 0)
dWa = Xb.T @ dz1
lr = 0.05

Wa = Wa - lr * dWa
Wb = Wb - lr * dWb
z1 = Xb @ Wa
h = np.maximum(0, z1)
y_hat = h @ Wb

loss_after = np.mean((y_hat - Yb)**2)

print("Loss after :", loss_after)
```

```python
X = np.random.randn(4, 3)
Y = np.random.randn(4, 1)
W1 = np.random.randn(3, 5) * 0.1
W2 = np.random.randn(5, 1) * 0.1

z1 = X @ W1
h  = np.maximum(0, z1)
y_hat = h @ W2
loss = ((y_hat - Y) ** 2).mean()
print('Initial loss:', round(loss, 4))
```

```python
dy   = 2 * (y_hat - Y) / Y.size
dW2  = h.T @ dy
dh   = dy @ W2.T
dz1  = dh * (z1 > 0)
dW1  = X.T @ dz1

print('dW1 shape:', dW1.shape, '(matches W1)')
print('dW2 shape:', dW2.shape, '(matches W2)')
```

```python

lr = 0.1
W1 -= lr * dW1
W2 -= lr * dW2
h_new = np.maximum(0, X @ W1)
loss_new = ((h_new @ W2 - Y) ** 2).mean()
print('Loss before:', round(loss, 4))
print('Loss after :', round(loss_new, 4), '-> should be lower')
```

```python
import sympy as sp

x, y = sp.symbols('x y')

f = x**4 + y**2

# Gradient
gradient = sp.Matrix([
    sp.diff(f, x),
    sp.diff(f, y)
])

# Hessian
hessian = gradient.jacobian([x, y])

print("Gradient:")
print(gradient)

print("Hessian:")
print(hessian)
```
