```python
# Part F – Calculus: Differentiation
# Suppose the predicted final score is represented by the function:
# F(x) = 40 + 5x − 0.2x²
# where x represents study hours.
# Task 6:
# 20. Find dF/dx manually.
# 21. Show that F′(x) = 5 − 0.4x.
# 22. Find the value of x for which F′(x) = 0.
# 23. Interpret the result in the context of study hours.
# Suggested Python verification:
# def F(x):
# return 40 + 5*x - 0.2*x**2
# def dF(x):
# return 5 - 0.4*x
# for x in range(1, 11):
# print(x, F(x), dF(x))

def F(x):
    return 40 + 5*x - 0.2*x**2

def dF(x):
    return 5 - 0.4*x

print("F(x) = 40 + 5x - 0.2x²")
print("F'(x) = 5 - 0.4x")

x = 5 / 0.4

print("\nValue of x when F'(x) = 0:")
print(x)

print("\nInterpretation:")
print("The derivative represents the rate of change of predicted final score with respect to study hours.")
print("The model reaches its maximum predicted value at 12.5 study hours.")

print("\nStudy Hours | F(x) | F'(x)")

for x in range(1, 11):
    print(x, "|", F(x), "|", dF(x))
```

```python
# Part G – Calculus: Integration
# Suppose the learning benefit is represented by:
# L(x) = 5x − 0.2x²
# Task 7: Calculate ∫₀⁵ L(x) dx.

# 24. Solve the integral manually.
# 25. Implement a numerical approximation using Python.
# 26. Explain what the area under the curve represents in this case.


x_values = np.linspace(0, 5, 1000)
y_values = 5*x_values - 0.2*x_values**2

integral_value = 2.5 * (5**2) - (0.2 / 3) * (5**3)

print("L(x) = 5x - 0.2x²")

print("\nManual Integration:")
print("∫₀⁵ L(x) dx =", integral_value)

numerical_integral = np.trapezoid(y_values, x_values)

print("\nNumerical Approximation:")
print(numerical_integral)

plt.figure(figsize=(8, 5))
plt.plot(x_values, y_values)
plt.fill_between(x_values, y_values, alpha=0.3)
plt.xlabel("Study Hours")
plt.ylabel("Learning Benefit")
plt.title("Learning Benefit Curve")
plt.grid(True)
plt.show()

print("\nThe area under the curve represents the total learning benefit accumulated between 0 and 5 study hours.")
```

```python
# Part I – Mini Prediction Model Using Linear Algebra
# Construct a simple linear model:
# Y = b₀ + b₁X₁ + b₂X₂ + b₃X₃
# where Y = Final Score, X₁ = Study Hours, X₂ = Attendance and X₃ = Assignment Score.
# Using matrix notation:
# Y = Xβ
# Estimate the coefficients using:
# β = (XᵀX)⁻¹XᵀY
# Suggested NumPy implementation:
# X = df[["Study_Hours","Attendance", "Assignment_Score"]].values
# Y = df["Final_Score"].values
# X = np.column_stack((np.ones(len(X)), X))
# beta = np.linalg.inv(X.T @ X) @ X.T @ Y
# print("Coefficients:)
# print(beta)
# Students should explain, in simple words, what the estimated coefficients indicate about the relationship
# between the input variables and final score.


X = df[["Study_Hours", "Attendance", "Assignment_Score"]].values
Y = df["Final_Score"].values

X_with_intercept = np.column_stack((np.ones(len(X)), X))

beta = np.linalg.inv(
    X_with_intercept.T @ X_with_intercept
) @ X_with_intercept.T @ Y

print("Estimated Coefficients:")

print("b0 (Intercept):", beta[0])
print("b1 (Study Hours):", beta[1])
print("b2 (Attendance):", beta[2])
print("b3 (Assignment Score):", beta[3])

print("\nPrediction Equation:")

print(
    "Final Score =",
    round(beta[0], 4),
    "+",
    round(beta[1], 4),
    "* Study Hours +",
    round(beta[2], 4),
    "* Attendance +",
    round(beta[3], 4),
    "* Assignment Score"
)

print("\nInterpretation:")
print("b0 represents the estimated intercept.")
print("b1 represents the relationship between study hours and final score.")
print("b2 represents the relationship between attendance and final score.")
print("b3 represents the relationship between assignment score and final score.")
```
