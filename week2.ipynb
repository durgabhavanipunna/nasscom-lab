```python
import os
import datetime
import random

print("--- Intermediate Python Core Exercise ---")
print("This exercise covers Python fundamentals, data structures, functions,")
print("core libraries (os, datetime, random), file handling, and exception handling.")
print("Estimated completion time: 3-4 hours.")
print("------------------------------------------")

# --- Global Data for the Exercise (Simulated Data) ---
DEVICE_TYPES = ['Temperature_Sensor', 'Humidity_Sensor', 'Pressure_Sensor', 'Light_Sensor']
LOCATIONS = ['Lab1', 'Lab2', 'OfficeA', 'OfficeB', 'Warehouse']
READING_RANGES = {
    'Temperature_Sensor': (18.0, 30.0),
    'Humidity_Sensor': (30.0, 80.0),
    'Pressure_Sensor': (980.0, 1030.0),
    'Light_Sensor': (100, 1000)
}

print("\n--- Task 1: Data Generation and Basic Structures ---")

# --- Task 1.1: Generate Raw Simulated Data ---
# Creating 50 random sensor readings from the past 24 hours
raw_sensor_data = []
base_time = datetime.datetime.now()

for i in range(50):
    device_type = random.choice(DEVICE_TYPES)
    min_val, max_val = READING_RANGES[device_type]

    # Generate random timestamp within the last 24 hours
    random_minutes = random.randint(0, 1440)
    timestamp = base_time - datetime.timedelta(minutes=random_minutes)

    reading = {
        'timestamp': timestamp.strftime("%Y-%m-%d %H:%M:%S"),
        'device_id': f"DEV_{device_type[:3].upper()}_{i:03d}",
        'device_type': device_type,
        'location': random.choice(LOCATIONS),
        'value': round(random.uniform(min_val, max_val), 2)
    }
    raw_sensor_data.append(reading)

# --- Task 1.2: Filter and Transform (Your Code) ---
temperature_readings = [
    {
        'timestamp': data['timestamp'],
        'device_id': data['device_id'],
        'location': data['location'],
        'temperature_celsius': data['value']
    }
    for data in raw_sensor_data
    if data['device_type'] == 'Temperature_Sensor'
]

# --- Task 1.3: Indexing, Slicing & Verification ---
print(f"Total raw records generated: {len(raw_sensor_data)}")
print(f"Total temperature records filtered: {len(temperature_readings)}")

if temperature_readings:
    print("\nFirst 3 temperature readings (Slicing):")
    for read in temperature_readings[:3]:
        print(read)
else:
    print("\nNo temperature readings were generated in this random batch.")
```
